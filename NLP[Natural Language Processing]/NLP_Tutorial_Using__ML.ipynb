```python
from sklearn.preprocessing import LabelEncoder

encode = LabelEncoder()

df['emotion'] = encode.fit_transform(df['emotion'])
```

# data cleaning!!

```python
# lowercasing every text-
df['text']=df['text'].apply(lambda x: x.lower())
df.head()
```

```python
# remove punctuation !!
import string

def remove_punc(txt):
  return txt.translate(str.maketrans('','',string.punctuation))


df['text'] = df['text'].apply(remove_punc)
df.head()
```

```python
# remove digit !!
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

df['text'] = df['text'].apply(remove_numbers)
df.head()
```

```python
# remove url & links !!
import re

def remove_urls(text):
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['text']=df['text'].apply(remove_urls)
df.head()
```

```python
# remove emojis from text !!
def remove_emojis(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new

df['text'] = df['text'].apply(remove_emojis)
```

```python
# remove stopwords !!
import nltk


from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
stop_words = set(stopwords.words('english'))
len(stop_words)
```

```python
df.loc[1]['text']
```

```python
def remove(txt):
  words = word_tokenize(txt)
  cleaned = []
  for i in words:
    if not i in stop_words:
      cleaned.append(i)

  return ' '.join(cleaned)

df['text'] = df['text'].apply(remove)
```

```python
df.loc[1]['text']
```

```python
# bag of words!
# CountVectorizer & TfidfVectorizer se hum data cleaning ke sare process kr sakte hai!!
from sklearn.feature_extraction.text import CountVectorizer

doc=[
     'this is the first document',
     'this document is the second document',
     'and this is the third one',
     'is this the first document'
]

vectorizer=CountVectorizer(ngram_range=(2,2))
X= vectorizer.fit_transform(doc)

print('vocabulary :',vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
# TF-IDF  [Term Frequency & Inverse Document Frequency]

from sklearn.feature_extraction.text import TfidfVectorizer

doc=[
     'this is the first document',
     'this document is the second document',
     'and this is the third one',
     'is this the first document'
]

vectorizer=TfidfVectorizer()
X= vectorizer.fit_transform(doc)

print('vocabulary :',vectorizer.get_feature_names_out())
print(X.toarray().round(2))
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split( df['text'], df['emotion'] ,test_size=0.2,random_state=42)
```

```python
from sklearn.feature_extraction.text import CountVectorizer , TfidfVectorizer

bow_vectorizer = CountVectorizer()
tfidf_vectorizer = TfidfVectorizer()

X_train_BOW = bow_vectorizer.fit_transform(X_train)
X_test_BOW = bow_vectorizer.transform(X_test)

X_train_TFIDF = tfidf_vectorizer.fit_transform(X_train)
X_test_TFIDF = tfidf_vectorizer.transform(X_test)
```

# MULTINOMINAL NAIVE BAYES

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

mnb_model = MultinomialNB()
mnb_model.fit(X_train_BOW,y_train)      #Multinominal naive bayes model made by dataset using BOW
```

```python
bow_predict = mnb_model.predict(X_test_BOW)
bow_predict
```

```python
accuracy_score(y_test,bow_predict)
```

```python
mnb_model.fit(X_train_TFIDF,y_train)         #Multinominal naive bayes model made by dataset using TFIDF
```

```python
tfidf_predict = mnb_model.predict(X_test_TFIDF)
tfidf_predict
```

```python
accuracy_score(y_test,tfidf_predict)
```
