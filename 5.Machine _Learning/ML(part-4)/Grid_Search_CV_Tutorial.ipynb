```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset('iris')
df
```

```python
df['species'].unique()
```

```python
X=df.drop('species',axis=1)
y=df['species']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

# parameter are affecting model accuracy directly!!

#KNN

```python
from sklearn.neighbors import KNeighborsClassifier

knn_model=KNeighborsClassifier(n_neighbors=3)
knn_model.fit(X_train,y_train)
```

```python
knn_model.score(X_test,y_test)             # n_neighbors=3
```

```python
knn_model=KNeighborsClassifier(n_neighbors=13)
knn_model.fit(X_train,y_train)
```

```python
knn_model.score(X_test,y_test)              # n_neighbors=13
```

# SVM

```python
from sklearn.svm import SVC

svm_model = SVC(C=30,gamma= 'auto',kernel='rbf')
svm_model.fit(X_train,y_train)
```

```python
svm_model.score(X_test,y_test)               # C=30
```

```python
svm_model = SVC(C=10,gamma= 'auto',kernel='rbf')
svm_model.fit(X_train,y_train)
```

```python
svm_model.score(X_test,y_test)                # C=10
```

# now lets use the grid search CV on SVM

```python
# cv stands for Cross Validation-
```

```python

svm_model = SVC(gamma= 'auto')
svm_model.fit(X_train,y_train)
```

```python
from sklearn.model_selection import GridSearchCV

classifier = GridSearchCV((svm_model),{
    'C' : [1,10,20,30],
    'kernel' :['rbf','linear'],
},cv = 5,return_train_score = False)
```

```python
classifier.fit(X,y)
```

```python
classifier.cv_results_
```

```python
result=pd.DataFrame(classifier.cv_results_)
result
```

```python
result[['param_C','param_kernel','mean_test_score']]
```

# Random searchCV on SVM






```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
classifier_r = RandomizedSearchCV((svm_model),{
    'C' : [1,10,20,30],
    'kernel' :['rbf','linear'],
},n_iter = 4,cv = 5,return_train_score = False)
```

```python
classifier_r.fit(X,y)
```

```python
results = pd.DataFrame(classifier_r.cv_results_)
results
```

```python
results[['param_C','param_kernel','mean_test_score']]
```
