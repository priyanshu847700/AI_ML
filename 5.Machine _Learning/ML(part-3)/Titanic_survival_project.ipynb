```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings('ignore')
```

```python
df=sns.load_dataset('titanic')
```

```python
df.head()
```

```python
df.columns
```

```python
df.shape
```

```python
df.info()
```

```python
df.drop(['alive', 'class', 'who', 'adult_male', 'embark_town', 'deck'],inplace=True,axis=1)
```

```python
df
```

```python
df.info()
```

```python
df["age"].fillna(df["age"].mean(), inplace=True)
```

```python
df.dropna(subset=["embarked"], inplace=True)
```

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
encode=LabelEncoder()

df['sex']=encode.fit_transform(df['sex'])
df['embarked']=encode.fit_transform(df['embarked'])
```

```python
df.head()
```

```python
df=df.astype(int)
df
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()

df[['age', 'fare']]=scaler.fit_transform(df[['age', 'fare']])
```

```python
df.head()
```

```python
X=df.drop('survived',axis=1)
y=df['survived']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

# using **logestic regression**

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()

model.fit(X_train,y_train)
```

```python
y_predict=model.predict(X_test)
y_predict
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix , classification_report
```

```python
confusion_matrix(y_test,y_predict)
```

```python
accuracy_score(y_test,y_predict)
```

```python
print(classification_report(y_test,y_predict))
```

# using KNN - Kth nearest **neighbour**

```python

from sklearn.neighbors import KNeighborsClassifier
knn_model=KNeighborsClassifier(n_neighbors=5)

knn_model.fit(X_train,y_train)
```

```python
knn_predict=knn_model.predict(X_test)
knn_predict
```

```python
y_test
```

```python
accuracy_score(y_test,knn_predict)                            # accuracy !!
```

```python
confusion_matrix(y_test,knn_predict)
```

```python
print(classification_report(y_test,knn_predict))
```

```python
from sklearn.naive_bayes import GaussianNB
native_bayes_model=GaussianNB()

native_bayes_model.fit(X_train,y_train)
```

```python
native_bayes_predict=native_bayes_model.predict(X_test)
native_bayes_predict
```

```python
y_test
```

```python
accuracy_score(y_test,native_bayes_predict)
```

```python
confusion_matrix(y_test,native_bayes_predict)
```

```python
print(classification_report(y_test,native_bayes_predict))
```

Using **DESIGION TREE**

```python
from sklearn.tree import DecisionTreeClassifier
decision_tree_model=DecisionTreeClassifier(random_state=42)

decision_tree_model.fit(X_train,y_train)
```

```python
decision_tree_predict=decision_tree_model.predict(X_test)
decision_tree_predict
```

```python
y_test
```

```python
accuracy_score(y_test,decision_tree_predict)
```

```python
confusion_matrix(y_test,decision_tree_predict)
```

```python
print(classification_report(y_test,decision_tree_predict))
```

Using SVM **[SUPPORT VECTOR MACHINE]**

```python
from sklearn.svm import SVC
svm_model=SVC(kernel='rbf')

svm_model.fit(X_train,y_train)
```

```python
svm_predict=svm_model.predict(X_test)
svm_predict
```

```python
y_test
```

```python
accuracy_score(y_test,svm_predict)
```

```python
confusion_matrix(y_test,svm_predict)
```

```python
print(classification_report(y_test,svm_predict))
```
