# Pandas Series
- A Series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index.

# Creating Series
- There are multiple ways to create a pandas Series:

```python
import pandas as pd
```

```python
import numpy as np
```

```python
labels=['a','b','c']
list=[1,2,3]
arr=np.array([4,5,6])
dict={1:10,2:20,3:30}
```

```python
pd.Series(list,index=labels)
```

```python
pd.Series(arr)
```

```python
pd.Series(dict)
```
