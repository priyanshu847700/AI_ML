# deep & shallow copy

```python
a=np.arange(1,21)
```

```python
slice=a[:5]
slice
```

```python
slice+10
```

```python
a                                                 # no change in a 
```

```python
b=a                                                            #so always use a.copy() -> fxn to make shallow copy
b[0]=99
```

```python
print(b)
print(a)
```

```python
b=a.copy()
b[1]=20
```

```python
print(b)
print(a)
```

# matrix operations!

```python
a=np.arange(1,5).reshape(2,2)
a
```

```python
b=np.arange(5,9).reshape(2,2)
b
```

```python

print(a @ b)                                               # dot product          ->        @
print(np.dot(a,b))
```

```python
 a.T                                                       # transpose of a matrix     ->    T
```
