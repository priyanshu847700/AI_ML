```python
import numpy as np;
```

# numpy indexing & slicing of vector

```python
arr=np.arange(11,21)
arr
```

```python
arr[4]
```

```python
print(arr[0:5])
print(arr[:5])
```

```python
print(arr[:])
print(arr[1::2])
```

# numpy indexing & slicing of matrix

```python
arr=np.arange(1,31).reshape(6,5);
arr
```

```python
arr[1]
```

```python
arr[1][0]
```

```python
arr[0:2,1:3]
```

```python
arr[3:6,3:6]
```

```python
arr[:,0]
```

# boolean indexing

```python
arr=np.arange(11,21)
arr
```

```python
arr[arr%2==0]
```
