```python
import seaborn as sns
import pandas as pd
import cufflinks as cf
import cufflinks.colors as colors
import cufflinks.plotlytools as pt
from plotly.offline import iplot

# Save original function
original_to_rgba = colors.to_rgba

# Fix NumPy float64 alpha
def fixed_to_rgba(color, alpha):
    return original_to_rgba(color, float(alpha))

# Patch both places
colors.to_rgba = fixed_to_rgba
pt.to_rgba = fixed_to_rgba

cf.go_offline()
```

```python
tips=sns.load_dataset('tips')
tips
```

```python
tips['total_bill'].iplot()
```

```python
tips.groupby('day')['tip'].mean().iplot(kind = 'bar')
```

```python
tips.iplot(kind='scatter',x ='total_bill',y = 'tip',mode = 'markers')
```

```python
tips[['total_bill','day']].iplot(kind = 'box',x = 'total_bill',y = 'day')
```
