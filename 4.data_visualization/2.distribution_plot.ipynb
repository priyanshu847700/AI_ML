```python
import seaborn as sns;
import matplotlib.pyplot as plt;
```

# We use distribution plots to visualize the distribution of quantitative data. [quantitative data]

## Let's discuss some plots that allow us to visualize the distribution of a data set. These plots are:

- distplot/histplot
- jointplot
- pairplot
- rugplot

```python
df=sns.load_dataset('tips')                                    # https://github.com/mwaskom/seaborn-data
df
```

# distplot/histplot

```python

sns.displot(df['total_bill'],bins=100)              
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],kde=True)   #kde -> kernel density estimater
plt.subplot(1,2,2)
sns.histplot(df['tip'],kde=True)
```

# jointplot

```python
sns.jointplot(x="total_bill" , y="tip" ,data=df ,kind='scatter')
```

```python
sns.jointplot(x="total_bill" , y="tip" ,data=df ,kind='hist')
```

```python
sns.jointplot(x="total_bill" , y="tip" ,data=df ,kind='kde')
```

```python
sns.jointplot(x="total_bill" , y="tip" ,data=df ,kind='hex')
```

```python
sns.jointplot(x="total_bill" , y="tip" ,data=df ,kind='reg')
```

```python
sns.jointplot(x="total_bill" , y="tip" ,data=df ,kind="resid")
```

```python
# pairplot
```

```python
sns.pairplot(df,hue='sex',palette="rainbow")
```

```python
sns.pairplot(df,hue='size',palette="rainbow")
```

```python
 # rugplot
```

```python
sns.rugplot(df)
```
