```python
import seaborn as sns 
import matplotlib.pyplot as plt 
```

# Heat map

```python
flights = sns.load_dataset('flights')                             # https://github.com/mwaskom/seaborn-data/blob/master/flights.csv
flights
```

```python
tips = sns.load_dataset("tips")
tips
```

```python
tipscorr  = tips[['total_bill','tip','size']]
```

```python
tipscorr
```

```python
tipscorr.corr()                         # co-relation
```

```python
sns.heatmap(tipscorr.corr())
```

```python
sns.heatmap(tipscorr.corr(),annot= True)
```

# Cluster map

```python
sns.clustermap(tipscorr.corr())
```

# pivot table Heat map

```python
pvtflight = flights.pivot_table(values='passengers',index='month',columns='year')
pvtflight
```

```python
sns.heatmap(pvtflight)
```
