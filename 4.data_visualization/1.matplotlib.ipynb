```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

# Matplotlib Basics

```python
x=np.linspace(0,5,11);
x
```

```python
y= x**2
y
```

```python
plt.plot(x,y)
plt.title("plot tutorial")
plt.xlabel("X - axis ")
plt.ylabel("Y - axis ")
```

# Subplots in Matplotlib

```python
plt.subplot(2,2,1)
plt.plot(y,y**3)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(x,x**3)
```

# Object Oriented way

- Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keyword arguments.

- figsize is a tuple of the width and height of the figure in inches dpi is the dots-per-inch (pixel per inch).

```python
fig=plt.figure(figsize=(10,5),dpi=100);
axes1=fig.add_axes([0.1,0.1,.8,.8])
axes1.plot(x,y)
axes2=fig.add_axes([0.2,0.4,.3,.3])
axes2.plot(x,y)
```
